### 📊 Scenario:
You’re helping an athletic director create a quick dashboard for just **five sports**. She wants to better understand how men’s and women’s revenue compare across a few programs. Your job is to make a simple Dash app with a pie chart or scatter plot showing the difference.

---
### ✅ Your Task (Step-by-Step)

- Load the sports.csv file using pandas.

- Drop missing values in the sports, rev_men, and rev_women columns.

- Pick 5 unique sports (your choice!) and filter the DataFrame to only include those.

    - Example: "Basketball", "Tennis", "Soccer", "Volleyball", "Golf"

- Create a new column called "Total_Revenue" by adding men’s and women’s revenue.

- Create either a pie chart or a scatter plot:

```python
# Import packages 

import dash
from dash import html, dcc
import pandas as pd
import plotly.express as px
import dash_bootstrap_components as dbc
```

```python
# Load and filter the data
df = pd.read_csv('/Users/Marcy_Student/Desktop/marcy/DA2025_Lectures/Mod2/data/sports.csv')
df = df[["sports", "rev_men", "rev_women"]].dropna()
```

```python
# Pick 5 sports
top5 = ["Basketball", "Tennis", "Soccer", "Volleyball", "Golf"]
#Copying the dataframe to not overwrite the original 
df_5 = df[df["sports"].isin(top5)].copy()
df_5
```

```python
# Create new column called Total_Revenue that adds up the men and women's revenue columns
df_5["Total_Revenue"] = df['rev_men'] + df['rev_women']
df_5
```

```python
# Make your pie or scatteplot using plotly 

fig = px.pie(df_5, names='sports', hole=0.2, color='sports',title='Sports repartition',template='plotly_dark')
fig.show()
```

```python
# Make the App -- DO NOT RUN THIS CELL YET It may give you a "port already in use error if you do"

from dash import Input, Output


app = dash.Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])
app.title = 'MY DASH APP'

app.layout = dbc.Container([
    dbc.Row(
        dbc.Col(html.H1(children='Revenue Analysis for 5 Sports'), class_name='text-center bg-light my-5', width=10)),
    dbc.Row([
        dbc.Col(dcc.Graph(figure=fig,id='my_pie',className='bg-light d-flex'), width=12)]),
    dbc.Row([
        dbc.Col(html.Button('JUST DO IT', id='My_button', className='btn-lg my-3'),width='4'),
        dbc.Col(dcc.Slider(0, 4, marks={i: f'Label{i}' for i in range(5)}, value=3))
    ]),
    dbc.Row([
        dbc.Col(dcc.Dropdown(
            id='Sports',
            options=[{'label':S, 'value':S} for S in sorted(df_5['sports'].unique())],
            placeholder='select a sport',
            style={'width':'50%'}
        ))
    ])
])

@app.callback(
    Output('my_pie', 'figure'),
    Input('Sports','value')
)

def update_chart(my_sport):
    filtered = df_5[df_5["sports"]== my_sport] if my_sport else df_5
    fig = px.pie(filtered, names='sports', hole=0.2, color='sports',title='Sports repartition',template='plotly_dark')
    return fig
if __name__ == '__main__':
    app.run(debug=True)

```
